```python
for num in range(1, 51):
    if num > 1:
        is_prime = True

        for i in range(2, int(num ** 0.5) + 1):
            if num % i == 0:
                is_prime = False
                break

        if is_prime:
            print(num, end=" ")
```

```python
plt.figure(figsize=(8,5))
plt.plot(df1["Student_ID"],df1["Maths"],marker='o',linestyle='--',linewidth=2)
plt.title("Maths Marks")
plt.xlabel("Student_ID")
plt.ylabel("Marks")
plt.grid(True)
plt.show()
```

```python
plt.figure(figsize=(15,7))
plt.plot(df1["Student_ID"],df1["Maths"],label='Maths',marker='o')
plt.plot(df1["Student_ID"],df1["English"],label='English',marker='s')
plt.plot(df1["Student_ID"],df1["Science"],label='Science',marker='^')
plt.title("Marks")
plt.xlabel("Student_ID")
plt.ylabel("Marks")
plt.grid(True)
plt.legend()
plt.show()
```
