```python
import pandas as pd
import numpy as np
fun=pd.read_csv('/startup_funding.csv')
np.random.seed(0)
```

```python
fun.head()
```

```python
fun.isnull()
```

```python
missing=fun.isnull().sum()
missing
```

```python
total_cells=np.prod(fun.shape)
total_missing=missing.sum()
percent_missing=(total_missing/total_cells)*100
print(percent_missing)
```

```python
fun.drop('Unnamed: 9',axis=1,inplace=True)
```

```python
fun =fun.rename(columns={
    'Date dd/mm/yyyy':'Date',
    'City  Location':'City',
    'InvestmentnType':'Investment_Type',
    'Amount in USD':'Amount_USD'
    })
```

```python
fun.fillna(0)
```

```python
fun['Date']=pd.to_datetime(fun['Date'],errors='coerce')
```

```python
fun
```

```python
print(fun['Amount_USD'].dtype)
```

```python
fun['Amount_USD'] = (
    fun['Amount_USD']
    .astype(str)
    .str.replace(',', '', regex=False)
)

fun['Amount_USD'] = pd.to_numeric(fun['Amount_USD'], errors='coerce')
```

```python
print(fun['Amount_USD'].dtype)
```

```python
# Industry Vertical
fun['Industry Vertical'].fillna('Unknown', inplace=True)

# SubVertical
fun['SubVertical'].fillna('Not Specified', inplace=True)

# City
fun['City'].fillna('Unknown', inplace=True)

# Investors Name
fun['Investors Name'].fillna('Undisclosed', inplace=True)
```

```python
fun
```

```python
fun['Investment_Type'].value_counts()
```

```python
fun['Investment_Type'] = fun['Investment_Type'].str.strip()
```

```python
fun['Investment_Type'] = fun['Investment_Type'].replace({
    'Seed/ Angel Funding': 'Seed Funding',
    'Seed / Angel Funding': 'Seed Funding',
    'Seed/Angel Funding': 'Seed Funding',
    'Seed / Angle Funding': 'Seed Funding',
    'Seed\\nFunding': 'Seed Funding',
    'Angel / Seed Funding': 'Seed Funding',
    'Seed Round': 'Seed Funding',
})
```

```python
fun['Investment_Type'].value_counts()
```

```python
print(fun.duplicated().sum())
```

```python
fun['Startup Name'] = fun['Startup Name'].str.strip()
```

```python
fun['Startup Name'].value_counts()
```

```python
fun['Funding_Category'] = pd.cut(
    fun['Amount_USD'],
    bins=[0,1000000,10000000,50000000,1000000000],
    labels=['Small','Medium','Large','Mega']
)
```

```python
print(fun['Funding_Category'].value_counts())
```

```python
fun.to_csv('startup_funding_cleaned.csv', index=False)
```
